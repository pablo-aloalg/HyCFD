```python
from stl import mesh
import numpy as np

# small thickness so snappyHexMesh can detect the surface
z0 = 0.0
z1 = 0.1

# terrain profile (x,y)
profile = np.array([
    [0, -10],
    [200, -10],
    [550, 8],
    [550, 8]
])

faces = []

for i in range(len(profile)-1):

    p0 = profile[i]
    p1 = profile[i+1]

    v0 = [p0[0], p0[1], z0]
    v1 = [p1[0], p1[1], z0]
    v2 = [p1[0], p1[1], z1]
    v3 = [p0[0], p0[1], z1]

    # two triangles per quad
    faces.append([v0, v1, v2])
    faces.append([v0, v2, v3])

faces = np.array(faces)

# create STL mesh
terrain = mesh.Mesh(np.zeros(faces.shape[0], dtype=mesh.Mesh.dtype))

for i, f in enumerate(faces):
    terrain.vectors[i] = f

terrain.save("bathymetry.stl")

print("bathymetry.stl created")
```

```python
from stl import mesh
import numpy as np

# small thickness so snappyHexMesh can detect the surface
z0 = 0.0
z1 = 0.1

# terrain profile (x,y)

depth_array = - np.loadtxt('inputs/templates/molokai/constant/triSurface/depth.bot')
x_array = np.arange(0,len(depth_array),1)
profile = np.column_stack((x_array, depth_array))

faces = []

for i in range(len(profile)-1):

    p0 = profile[i]
    p1 = profile[i+1]

    v0 = [p0[0], p0[1], z0]
    v1 = [p1[0], p1[1], z0]
    v2 = [p1[0], p1[1], z1]
    v3 = [p0[0], p0[1], z1]

    # two triangles per quad
    faces.append([v0, v1, v2])
    faces.append([v0, v2, v3])

faces = np.array(faces)

# create STL mesh
terrain = mesh.Mesh(np.zeros(faces.shape[0], dtype=mesh.Mesh.dtype))

for i, f in enumerate(faces):
    terrain.vectors[i] = f

terrain.save("inputs/templates/molokai/constant/triSurface/bathymetry.stl")

print("inputs/templates/molokai/constant/triSurface/bathymetry.stl created")
```
